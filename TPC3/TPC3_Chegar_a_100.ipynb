```python
import random                                          # Importa variaveis aleatórios. É necessário para o random.randint(1, 10)
# Função em que o computador joga primeiro. O computador deve ganhar sempre.
def computador_primeiro():                             
    total = 0                                          # Contador do total para manter a pontuação durante o jogo. Começa no zero 
    terminou = False                                   # Cria a condição que indica se o jogo terminou. Enquanto está como False significia que o jogo ainda não terminou
    print("\n=== Computador joga primeiro ===")        # Imprime a notificação de quem joga primeiro nesta modalidade.
    jogada_computador = 1                              # Como o computador joga primeiro, começa obrigatoriamente por somar 1. Para o computador ganhar obrigatoriamente.
    total = total + jogada_computador                  # Adiciona a jogada do computador ao total.
    print("\nComputador soma:", jogada_computador)     # Imprime a jogada do computador
    print("Total:", total)                             # Imprime o resultado do total da jogada do computador
    while terminou == False:                           # Enquanto o jogo ainda não terminou
        if total < 100:                                                                      # Verifica se o total ainda é inferior a 100.
            jogada_jogador = int(input("\nEscolhe um número entre 1 e 10: "))                # Converte o que o utilizador colocar como número em texto para um número inteiro
            while jogada_jogador < 1 or jogada_jogador > 10 or total + jogada_jogador > 100: # Se o jogador escolher um número que é abaixo de 1 ou maior que 10 ou cuja soma ao total fosse ultrapassar 100.
                print("Jogada inválida.")                                                    # Imprime como jogada inválida
                jogada_jogador = int(input("Escolhe um número entre 1 e 10: "))              # Volta a indicar as opções válidas
            total = total + jogada_jogador                                                   # Adiciona a jogada do jogador ao total
            print("Jogador soma:", jogada_jogador)                                           # Imprime o que o jogador escolheu
            print("Total:", total)                                                           # Imprime o total que resulta do que o jogador escolheu
            if total == 100:                                                                 # Se o total chegar a 100
                print("\nO jogador venceu!")                                                 # Imprime que o jogador venceu. Não deve acontecer nesta modalidade.
                terminou = True                                                              # Termina o ciclo ao tornar a condição terminou verdadeira (True)
        if terminou == False:                                                                # Se o jogo ainda não terminou
            jogada_computador = 11 - jogada_jogador                                          # Esta é a parte principal da estratégia vencedora do computador que escolhe sempre um número que, somado ao número do jogador, dá 11.
            if total + jogada_computador > 100:                                              # Se o total da jogada do computador for maior que 100
                jogada_computador = 100 - total                                              # O computador escolhe o número que é necessário para chegar a 100 no final sem ultrapassar 100.
            total = total + jogada_computador                                                # Retorna o total após o computador fazer a sua escolha
            print("\nComputador soma:", jogada_computador)                                   # Imprime a jogada do computador
            print("Total:", total)                                                           # Imprime o total a que se chega quando o computador faz a sua escolha
            if total == 100:                                                                 # Se o total chega a 100
                print("\nO computador venceu!")                                              # Imprime que o computador venceu. Nesta modalidade, o computador obrigatoriamente tem que vencer.
                terminou = True                                                              # Termina o ciclo ao tornar a condição terminou verdadeira (True)

# Função em que o utilizador joga primeiro.
def utilizador_primeiro():
    total = 0
    terminou = False
    print("\n=== Utilizador joga primeiro ===")                          # Ao contrário da modalidade anterior, como é o utilizador que começa, não se diz que o computador começa em 1 para não garantir que ganha.
    while terminou == False:
        jogada_jogador = int(input("\nEscolhe um número entre 1 e 10: "))
        while jogada_jogador < 1 or jogada_jogador > 10 or total + jogada_jogador > 100:    
            print("Jogada inválida.")                      
            jogada_jogador = int(input("Escolhe um número entre 1 e 10: "))
        total = total + jogada_jogador
        print("Jogador soma:", jogada_jogador)
        print("Total:", total)
        if total == 100:
            print("\nO jogador venceu!")
            terminou = True
        if terminou == False:                                            # O computador só joga a seguir se o jogador ainda não tiver vencido.
            proximo_objetivo = 1                                         # Começa em 1 para seguir a estratégia da modalidade anterior para chegar a 100 antes do utilizador
            while proximo_objetivo <= total:                             # Enquanto o objetivo encontrado for menor ou igual ao total atual, significa que já passou do objetivo inicial e que tem de procurar o seguinte.
                proximo_objetivo = proximo_objetivo + 11                 # Avança para o número estratégico seguinte de 12, 23, 34, 45, 56, 67, 78, 89, 100. 
            jogada_computador = proximo_objetivo - total                 # Calcula quanto o computador precisa de acrescentar para chegar a esse objetivo.
            if jogada_computador >= 1 and jogada_computador <= 10:       # Se o número necessário está entre 1 e 10, porque só essas jogadas são permitidas.
                total = total + jogada_computador                        # Se a jogada estratégica for possível, o computador faz essa jogada.
            else:                                                        # Senão for possível chegar ao próximo número estratégico. Isto acontece precisamente quando o jogador conseguiu ficar numa posição estratégica.
                jogada_computador = random.randint(1, 10)                # O computador escolhe um número aleatório entre 1 e 10  
                while total + jogada_computador > 100:                   # Verifica se a jogada aleatória faria ultrapassar 100.
                    jogada_computador = random.randint(1, 10)            # Se ultrapassasse, gera outro número aleatório. Continua a fazer isso até encontrar uma jogada válida.
                total = total + jogada_computador                        # Quando encontra uma jogada válida, soma a sua jogada a esse total.
            print("\nComputador soma:", jogada_computador)
            print("Total:", total)
            if total == 100:                                             # Se o total do computador chega a 100
                print("\nO computador venceu!")                          # Imprime que o computador venceu.
                terminou = True                                          # Termina o ciclo.

print("=== Corrida para o 100 ===")                         # Imprime o nome do jogo
print("1 - Computador joga primeiro")                       # Imprime os nomes das modalidades a escolher
print("2 - Utilizador joga primeiro")                     
modalidade = input("Escolhe a modalidade (1 ou 2): ")       # Para escolher a modalidade pede ao utilizador para escolher entre 1 ou 2
while modalidade != "1" and modalidade != "2":              # Enquanto a modalidade escolhida for diferenete de 1 e 2
    print("Modalidade inválida.")                           # Imprime uma notificação a indicar uma escolha inválida
    modalidade = input("Escolhe a modalidade (1 ou 2): ")   # E dá de volta a opção das escolhas permitidas
if modalidade == "1":                                       # Se a modalidade for 1
    computador_primeiro()                                   # A função da modalidade 1 é chamada
else:                                                       # Senão:
    utilizador_primeiro()                                   # A função da modalidade 2 é chamada
```
