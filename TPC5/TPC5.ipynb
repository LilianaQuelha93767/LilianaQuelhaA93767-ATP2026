```python

# TPC5 - Sistema de Gestão de Alunos
# Modelo: aluno = (nome, aluno_id, [notaTPC, notaProj, notaTeste])
# Turma: lista de alunos

# Criar uma turma
def criarTurma():
    return []

# Inserir um aluno na turma
def inserirAluno(turma, nome, aluno_id, notas):
    aluno = (nome, aluno_id, notas)
    turma.append(aluno)
    print(f"Aluno '{nome}' inserido com sucesso.")
    return turma

# Listar a turma
def listarTurma(turma):
    print("\nLista de alunos na turma:")
    if not turma:
        print("A turma está vazia.")
    else:
        for aluno in turma:
            nome, aluno_id, notas = aluno
            notaTPC, notaProj, notaTeste = notas
            print(f"\nNome: {nome}")
            print(f"ID: {aluno_id}")
            print(f"Nota de TPC: {notaTPC}")
            print(f"Nota de Projeto: {notaProj}")
            print(f"Nota de Teste: {notaTeste}")

# Consultar um aluno através do ID
def consultarAluno(turma, aluno_id):
    encontrado = None
    for aluno in turma:
        if aluno[1] == aluno_id and encontrado is None:
            encontrado = aluno
    if encontrado is not None:
        nome, aluno_id, notas = encontrado
        notaTPC, notaProj, notaTeste = notas
        print("\nAluno encontrado:")
        print(f"Nome: {nome}")
        print(f"ID: {aluno_id}")
        print(f"Nota de TPC: {notaTPC}")
        print(f"Nota de Projeto: {notaProj}")
        print(f"Nota de Teste: {notaTeste}")
    else:
        print("Aluno não encontrado.")

    return encontrado


# Guardar a turma em ficheiro
def guardarTurma(turma, nome_arquivo):
    arquivo = open(nome_arquivo.txt, "w", encoding="utf-8")
    arquivo.write("Nome; ID; NotaTPC; NotaProjeto; NotaTeste\n")
    for aluno in turma:
        nome, aluno_id, notas = aluno
        notaTPC, notaProj, notaTeste = notas
        arquivo.write(f"{nome};{aluno_id};{notaTPC};{notaProj};{notaTeste}\n")
    arquivo.close()
    print("Turma guardada com sucesso.")


# Carregar uma turma num ficheiro 
def carregarTurma(nome_arquivo):
    turma = []
    arquivo = open(nome_arquivo.txt, "r", encoding="utf-8")
    for linha in arquivo:
        campos = linha.strip().split(";")
        if linha.strip() != "" and linha.strip() != "Nome; ID; NotaTPC; NotaProjeto; NotaTeste":
            nome = campos[0]
            aluno_id = campos[1]
            notaTPC = float(campos[2])
            notaProj = float(campos[3])
            notaTeste = float(campos[4])
            notas = [notaTPC, notaProj, notaTeste]
            aluno = (nome, aluno_id, notas)
            turma.append(aluno)
    arquivo.close()
    print("Turma carregada com sucesso.")
    return turma

# Menu interativo
def menu():
    turma = []
    executar = True
    while executar:
        print("\n--- Sistema de Gestão de Alunos ---")
        print("1: Criar uma turma")
        print("2: Inserir um aluno na turma")
        print("3: Listar a turma")
        print("4: Consultar um aluno por ID")
        print("5: Guardar a turma em ficheiro")
        print("6: Carregar uma turma dum ficheiro")
        print("0: Sair da aplicação")

        opcao = input("Selecione uma opção: ")

        if opcao == "1":
            turma = criarTurma()
            print("Nova turma criada com sucesso.")

        elif opcao == "2":
            nome = input("Insira o nome do aluno: ")
            aluno_id = input("Insira o ID do aluno: ")
            notaTPC = float(input("Insira a nota de TPC: "))
            notaProj = float(input("Insira a nota de Projeto: "))
            notaTeste = float(input("Insira a nota de Teste: "))
            notas = [notaTPC, notaProj, notaTeste]
            turma = inserirAluno(turma, nome, aluno_id, notas)

        elif opcao == "3":
            listarTurma(turma)

        elif opcao == "4":
            aluno_id = input("Insira o ID do aluno que deseja consultar: ")
            consultarAluno(turma, aluno_id)

        elif opcao == "5":
            nome_arquivo = input("Nome do ficheiro: ") or "turma.txt"
            guardarTurma(turma, nome_arquivo)

        elif opcao == "6":
            nome_arquivo = input("Nome do ficheiro: ") or "turma.txt"
            turma = carregarTurma(nome_arquivo)

        elif opcao == "0":
            print("A encerrar o sistema.")
            executar = False

        else:
            print("Opção inválida. Tente novamente.")

if __name__ == "__main__":
    menu()
```
