```python
import random

def jogador_adivinha():
    numero = random.randint(0, 100)
    tentativas = 0
    acertou = False

    print("\nO computador escolheu um número entre 0 e 100.")

    while acertou == False:
        palpite = int(input("Qual é o número em que pensei? "))

        if palpite >= 0 and palpite <= 100:
            tentativas = tentativas + 1

            if palpite == numero:
                print("Acertou!")
                acertou = True

            elif palpite < numero:
                print("O número que pensei é maior.")

            else:
                print("O número que pensei é menor.")

        else:
            print("O palpite deve estar entre 0 e 100.")

    print("Número de tentativas utilizadas:", tentativas)


def computador_adivinha():
    minimo = 0
    maximo = 100
    tentativas = 0
    acertou = False

    print("\nPensa num número entre 0 e 100.")
    print("O computador vai tentar adivinhar.")

    while acertou == False and minimo <= maximo:
        palpite = (minimo + maximo) // 2

        print("\nO computador escolheu:", palpite)
        print("1 - Acertou")
        print("2 - O número que pensei é maior")
        print("3 - O número que pensei é menor")

        resposta = input("Escolhe 1, 2 ou 3: ")

        while resposta != "1" and resposta != "2" and resposta != "3":
            print("Resposta inválida.")
            resposta = input("Escolhe 1, 2 ou 3: ")

        tentativas = tentativas + 1

        if resposta == "1":
            acertou = True
            print("Acertou!")

        elif resposta == "2":
            minimo = palpite + 1

        else:
            maximo = palpite - 1

    if acertou == True:
        print("Número de tentativas utilizadas:", tentativas)
    else:
        print("As respostas dadas não são compatíveis.")


print("=== Adivinha o Número ===")
print("1 - O computador pensa num número e o utilizador adivinha")
print("2 - O utilizador pensa num número e o computador adivinha")

opcao = input("Escolhe a modalidade (1 ou 2): ")

while opcao != "1" and opcao != "2":
    print("Modalidade incorreta.")
    opcao = input("Escolhe a modalidade (1 ou 2): ")

if opcao == "1":
    jogador_adivinha()
else:
    computador_adivinha()
```
