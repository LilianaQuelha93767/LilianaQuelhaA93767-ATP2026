```python
import random

lista = []

def menu():
    print("Menu")
    print("1) Criar Lista")
    print("2) Ler Lista")
    print("3) Soma")
    print("4) Média")
    print("5) Maior")
    print("6) Menor")
    print("7) estaOrdenada por ordem crescente")
    print("8) estaOrdenada por ordem decrescente")
    print("9) Procura um elemento")
    print("0) Sair")


def criaLista():
    n = int(input("Introduza o número de números da lista: "))
    lista = [random.randint(1, 101) for _ in range(n)]
    return lista


def leLista():
    N = int(input("Introduza o número de números da lista: "))
    lista = []
    i = 1
    while i <= N:
        lista.append(int(input(f"Introduza o número {i}: ")))
        i = i + 1
    return lista


def somaLista(lista):
    res = 0
    for num in lista:
        res = res + num
    return res


def mediaLista(lista):
    res = somaLista(lista) / len(lista)
    return res


def maiorLista(lista):
    res = lista[0]
    for num in lista:
        if num > res:
            res = num
    return res


def menorLista(lista):
    res = lista[0]
    for num in lista:
        if num < res:
            res = num
    return res

def estaOrdenadaC(lista):
    res = True
    i = 0
    while i < len(lista) - 1 and res:
        if lista[i] > lista[i + 1]:
            res = False

        i = i + 1
    return res


def estaOrdenadaD(lista):
    res = True
    i = 0
    while i < len(lista) - 1 and res:
        if lista[i] < lista[i + 1]:
            res = False
        i = i + 1
    return res


def procurar_elemento(lista, elem):
    res = -1
    i = 0
    while res == -1 and i < len(lista):
        if lista[i] == elem:
            res = i
        i = i + 1
    return res


menu()
opcao = input("Selecione a opção desejada: ")
while opcao != "0":
    if opcao == "1":
        lista = criaLista()
        print("A lista criada é a seguinte:", lista)

    elif opcao == "2":
        lista = leLista()
        print("A lista criada é a seguinte:", lista)

    elif opcao == "3":
        if lista:
            print("A soma resultante é a seguinte:", somaLista(lista))
        else:
            print("Por favor, crie uma lista primeiro.")

    elif opcao == "4":
        if lista:
            print("A média resultante é a seguinte:", mediaLista(lista))
        else:
            print("Por favor, crie uma lista primeiro.")

    elif opcao == "5":
        if lista:
            print("O maior elemento da lista é:", maiorLista(lista))
        else:
            print("Por favor, crie uma lista primeiro.")

    elif opcao == "6":
        if lista:
            print("O menor elemento da lista é:", menorLista(lista))
        else:
            print("Por favor, crie uma lista primeiro.")

    elif opcao == "7":
        if lista:
            if estaOrdenadaC(lista) == True:
                print("Sim.")
            else:
                print("Não.")
        else:
            print("Por favor, crie uma lista primeiro.")

    elif opcao == "8":
        if lista:
            if estaOrdenadaD(lista) == True:
                print("Sim.")
            else:
                print("Não.")
        else:
            print("Por favor, crie uma lista primeiro.")

    elif opcao == "9":
        if lista:
            elemento = int(input("Introduza o elemento que deseja procurar: "))
            posicao = procurar_elemento(lista, elemento)

            if posicao != -1:
                print(f"O elemento {elemento} está na posição {posicao}.")
            else:
                print(f"O elemento {elemento} não foi encontrado.")

        else:
            print("Crie uma lista primeiro!")

    else:
        print("Opção inválida. Por favor, tente outra vez.")

    menu()
    opcao = input("Selecione a opção desejada: ")

print("Lista atual:", lista)
print("Obrigado, volte sempre!")
```
